Flagging Vendor Invoices for Manual Review
Objective
To predict whether a vendor invoice needs manual checking based on unusual costs, freight charges, or delivery patterns. This will help reduce financial risks, save time, and make invoice checking more efficient.
Problem Statement
1. Checking every invoice manually takes a lot of time, especially when there are many invoices.

2. Unusual freight charges, price differences, or delivery delays may indicate errors or other problems.

3. An automated system can identify invoices that may need extra attention. This allows the finance team to focus on risky invoices while processing normal invoices more efficiently.
